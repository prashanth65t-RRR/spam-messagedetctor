```python
import os
from getpass import getpass

# Changed import for Google Gemini models
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage

from langgraph.graph import StateGraph, START, END
from typing import TypedDict


# API key (Changed to GOOGLE_API_KEY as the model is Gemini)
os.environ["GOOGLE_API_KEY"] = getpass("Enter Google API Key: ")


# State
class State(TypedDict):
    question: str
    answer: str


# LLM (Changed class to ChatGoogleGenerativeAI for Gemini models)
llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash", # Updated model from gemini-2.5-flash to gemini-3.6-flash
    temperature=0
)


# Agent
def agent(state):

    response = llm.invoke(
        [HumanMessage(content=state["question"])]
    )

    return {
        "answer": response.content
    }


# Create graph
graph = StateGraph(State)

graph.add_node("agent", agent)

graph.add_edge(START, "agent")
graph.add_edge("agent", END)

app = graph.compile()


# Run
question = input("Ask a question: ")

result = app.invoke({
    "question": question
})

print("\nAnswer:")
print(result["answer"])
```
